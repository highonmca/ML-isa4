```python
import numpy as np
import matplotlib.pyplot as plt
import os 
def generate_graphs():
    x = np.linspace(0, 10, 400)
    output_dir='function_graphs'
    os.makedirs(output_dir,exist_ok=True)
    functions = {
        'linear': lambda x: 2*x + 3,
        'quadratic': lambda x: x**2 - 4,
        'cubic': lambda x: 0.1*x**3 - x,
        'sin': lambda x: np.sin(x),
        'cos': lambda x: np.cos(x),
        'exponential': lambda x: np.exp(0.1 * x)
    }

    for name, func in functions.items():
        y = func(x)

        plt.figure(figsize=(6, 6))
        plt.axhline(0, color='black', linewidth=1)
        plt.axvline(0, color='black', linewidth=1)
        plt.grid(True, linestyle='--', alpha=0.6)

        plt.plot(x, y, label=f'{name} function')
        plt.legend()
        plt.xlabel('X-axis')
        plt.ylabel('Y-axis')
        plt.title(f'{name.capitalize()} Function Graph')

        plt.savefig(f'{output_dir}/{name}_graph.png')
        plt.close()
        print(f'Saved graphs as {name}_graph.png in the {output_dir} directory')

# Generate graphs
generate_graphs()
```
