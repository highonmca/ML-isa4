```python
import numpy as np
import matplotlib.pyplot as plt
import os

def generate_graphs(save_folder='graphs'):
    # Ensure the save folder exists
    os.makedirs(save_folder, exist_ok=True)

    x = np.linspace(0, 10, 400)
    y_min, y_max = -10, 10  # Set the same Y-axis range for all graphs

    functions = {
    # Polynomial Functions
    'linear': lambda x: 2*x + 3,
    'quadratic': lambda x: x**2 - 4,
    'cubic': lambda x: 0.1*x**3 - x,
    'quartic': lambda x: x**4 - 3*x**2 + 2,
    'quintic': lambda x: x**5 - 4*x**3 + x,

    # Trigonometric Functions
    'sin': lambda x: np.sin(x),
    'cos': lambda x: np.cos(x),
    'tan': lambda x: np.tan(x),
    'cot': lambda x: 1 / np.tan(x),  # cot(x) = 1/tan(x)
    'sec': lambda x: 1 / np.cos(x),  # sec(x) = 1/cos(x)
    'cosec': lambda x: 1 / np.sin(x),  # cosec(x) = 1/sin(x)

    # Exponential Functions
    'exponential': lambda x: np.exp(0.1 * x),  # e^(0.1x)
    'exp_2x': lambda x: np.exp(2 * x),  # e^(2x)
    'exp_neg_x': lambda x: np.exp(-x),  # e^(-x)

    # Logarithmic Functions (Avoid log(0) issue)
    'log': lambda x: np.log(np.abs(x) + 0.1),  # Natural log, avoiding log(0) issue
    'log10': lambda x: np.log10(np.abs(x) + 0.1),  # Base 10 log

    # Hyperbolic Functions
    'sinh': lambda x: np.sinh(x),
    'cosh': lambda x: np.cosh(x),
    'tanh': lambda x: np.tanh(x),

    # Rational Functions
    'rational_1': lambda x: (x**2 - 1) / (x**2 + 1),  # Example rational function
    'rational_2': lambda x: (x - 2) / (x + 3),

    # Piecewise Function (e.g., step function)
    'piecewise': lambda x: np.piecewise(x, [x < 0, x >= 0], [lambda x: -x, lambda x: x])
}

    for name, func in functions.items():
        y = func(x)

        plt.figure(figsize=(6, 6))
        plt.axhline(0, color='black', linewidth=1)
        plt.axvline(0, color='black', linewidth=1)
        plt.grid(True, linestyle='--', alpha=0.6)

        plt.plot(x, y, label=f'{name} function')
        plt.legend()
        plt.xlabel('X-axis')
        plt.ylabel('Y-axis')
        plt.title(f'{name.capitalize()} Function Graph')

        # Set fixed axis limits
        plt.xlim(0, 10)
        plt.ylim(y_min, y_max)

        filename = os.path.join(save_folder, f'{name}.png')
        plt.savefig(filename)
        plt.close()
        print(f'Saved {filename}')

# Generate graphs
generate_graphs()
```
